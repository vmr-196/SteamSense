# 03 - Data Preprocessing

## Objectives

- Clean games dataset
- Clean recommendations dataset
- Fix datatypes
- Handle missing values
- Remove duplicates
- Convert columns
- Save processed datasets

```python
import pandas as pd
import numpy as np

from pathlib import Path

pd.set_option("display.max_columns", None)
```

```python
RAW_PATH = Path("../data/raw")
PROCESSED_PATH = Path("../data/processed")

PROCESSED_PATH.mkdir(exist_ok=True)
```

```python
games = pd.read_csv(RAW_PATH / "games.csv")
recommendations = pd.read_csv(RAW_PATH / "recommendations.csv")
```

```python
games.info()
```

```python
recommendations.info()
```

```python
games.isnull().sum()
```

```python
recommendations.isnull().sum()
```

```python
print(games.duplicated().sum())
print(recommendations.duplicated().sum())
```

```python
games.drop_duplicates(inplace=True)
recommendations.drop_duplicates(inplace=True)
```

```python
recommendations["is_recommended"].unique()
recommendations["is_recommended"] = (
    recommendations["is_recommended"]
    .astype(bool)
)
```

```python
games["date_release"] = pd.to_datetime(
    games["date_release"]
)
recommendations["date"] = pd.to_datetime(
    recommendations["date"]
)
```

```python
games.info()
recommendations.info()
```

```python
games.to_csv(
    PROCESSED_PATH / "games_clean.csv",
    index=False
)

recommendations.to_csv(
    PROCESSED_PATH / "recommendations_clean.csv",
    index=False
)
```
