```python
import pandas as pd

from pathlib import Path
```

```python
RAW_PATH = Path("../data/raw")
PROCESSED_PATH = Path("../data/processed")
```

```python
metadata = pd.read_json(
    RAW_PATH / "games_metadata.json",
    lines=True
)
```

```python
metadata.head()
```

```python
metadata.info()
```

```python
metadata.columns
```

```python
games = pd.read_csv(
    PROCESSED_PATH / "games_features.csv"
)
```

```python
games = games.merge(
    metadata,
    on="app_id",
    how="left"
)
```

```python
games.head()
```

```python
games["tags"] = games["tags"].apply(
    lambda x: " ".join(x) if isinstance(x, list) else ""
)
```

```python
games["description"] = (
    games["description"]
    .fillna("")
)
```

```python
games["content"] = (
    games["title"] + " "
    + games["description"] + " "
    + games["tags"]
)
```

```python
games.to_csv(
    PROCESSED_PATH / "games_content.csv",
    index=False
)
```
