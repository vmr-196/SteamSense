```python
import pandas as pd
import numpy as np

from pathlib import Path

pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", None)
```

```python
DATA_PATH = Path("../data/processed")
```

```python
games = pd.read_csv(DATA_PATH / "games_clean.csv")
```

```python
games.head()
```

```python
games.columns.tolist()
```

```python
print(f"Rows : {games.shape[0]:,}")
print(f"Columns : {games.shape[1]}")
```

```python
games["date_release"] = pd.to_datetime(
    games["date_release"]
)
```

```python
games["release_year"] = games["date_release"].dt.year
```

```python
from datetime import datetime

CURRENT_YEAR = datetime.now().year

games["game_age"] = CURRENT_YEAR - games["release_year"]
```

```python
games["is_free"] = (
    games["price_final"] == 0
).astype(int)
```

```python
games["platform_count"] = (
    games["win"].astype(int)
    + games["mac"].astype(int)
    + games["linux"].astype(int)
)
```

```python
games["steam_deck"] = games["steam_deck"].astype(int)
```

```python
games["discount"] = games["discount"].astype(int)
```

```python
games["popularity_score"] = (
    games["positive_ratio"] * 0.6
    + np.log1p(games["user_reviews"]) * 40
)
```

```python
games["popularity_score"].describe()
```

```python
games.sort_values(
    by="popularity_score",
    ascending=False
)[
    [
        "title",
        "popularity_score",
        "positive_ratio",
        "user_reviews"
    ]
].head(10)
```

```python
rating_map = {
    "Overwhelmingly Negative": 1,
    "Very Negative": 2,
    "Mostly Negative": 3,
    "Negative": 4,
    "Mixed": 5,
    "Positive": 6,
    "Mostly Positive": 7,
    "Very Positive": 8,
    "Overwhelmingly Positive": 9
}
```

```python
games["rating_score"] = games["rating"].map(rating_map)
```
